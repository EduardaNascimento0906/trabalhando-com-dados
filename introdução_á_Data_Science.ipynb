Analizando as notas em geral

```python
import pandas as pd

tmdb = pd.read_csv("tmdb_5000_movies.csv")
tmdb.head()
```

```python
tmdb.original_language.unique()
```

```python
tmdb.original_language.value_counts() #devolve uma série
```

```python
tmdb.original_language.value_counts(). to_frame() #transforma em um dataframe
```

```python
contagem_de_lingua = tmdb["original_language"].value_counts().to_frame().reset_index()
contagem_de_lingua.columns = ["original_language", "total"]
contagem_de_lingua.head() #transformação de uma série em um dataframe
```

```python
sns.barplot(x="original_language", y="total", data=contagem_de_lingua)
```

```python
sns.catplot(x="original_language", kind="count", data=tmdb) #facilitando com seaborn
```

```python
plt.pie(contagem_de_lingua["total"], labels=contagem_de_lingua["original_language"])
#gráfico de pizza, pouco utilizando pois visualmente não é possivel facilmente analizar
#com dados desproporcionais em relação a quantidade como o apresentado
#criando um "pac man"
```

```python
total_por_lingua = tmdb["original_language"].value_counts()
total_geral = total_por_lingua.sum()
total_de_ingles = total_por_lingua.loc["en"]
total_do_resto = total_geral - total_de_ingles
print(total_do_resto)
total_por_lingua.loc["en"]
#separando dados para realizar uma comparação,
#assim poder criar uma visualização mais eficiente.
```

```python
dados = {
    'lingua' : ["ingles", "outros"],
    'total' : [total_de_ingles, total_do_resto]
}
dados
```

```python
pd.DataFrame(dados) #criando um novo dataframe
```

```python
sns.barplot(x="lingua", y="total", data = dados)
#criando uma visualização mais facil de ser interpretada
#mostrando que a escolha do tipo de gráfico influencia na forma de ver as informações
```

```python
plt.pie(dados["total"], labels=dados["lingua"])
#gráfico de pizza podendo ser melhor utilizado nos dados separados
#porem apenas para realizar uma análise visual, não quantitativa
#não consegue-se identificar valores númericos
```

```python
total_por_lingua_de_outros_filmes = tmdb.query("original_language != 'en'").original_language.value_counts()
total_por_lingua_de_outros_filmes
```

```python
filmes_sem_lingua_original_em_ingles = tmdb.query("original_language != 'en'")

sns.catplot(x="original_language", kind="count", data = filmes_sem_lingua_original_em_ingles,)
#criando uma visualização de dados com outras linguas fora o inglês
```

```python
filmes_sem_lingua_original_em_ingles = tmdb.query("original_language != 'en'")

sns.catplot(x="original_language", kind="count", data = filmes_sem_lingua_original_em_ingles, aspect = 2, order = total_por_lingua_de_outros_filmes.index, palette="hls")
#criando uma visualização de dados com outras linguas fora o inglês
#mais organizada visualmente e mais adptadae refinada para visualizar os dados.
```
